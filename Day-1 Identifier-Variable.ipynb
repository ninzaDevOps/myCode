# Day-1  Identifier -Variable
### A name use to identify a program element such as variable, function , class methods etcs.
### Variable: A name that refer to an object in memory during execution.

## Rules to definea variable or Identifier:
### Use letter, digits & underscore_
### Never Start identifier with digit.
### Do not use python keyword as a identifier ex : if, for etcs.
### Do not use space or hyphen- in identifier
### better to use snake_case ,db_name , SERVER_NAME,
### Avoid using Python Built-in functions as name variable name.

## Built-ins: Built-ins function are ready to use(pre defined)

```python
import builtins
print(dir(builtins))
```

```python
#built-in example
len=4
print(len)
#when we use built-in no error, but better avoid(not to use)
```

```python
len=4
print(len)
print(len("FDE"))
# Its not hard rule but better practice to avoid conflicts with built-ins , Dont use built-ins as a variable.
```

```python
#We can test keywords
import keyword
print(keyword.iskeyword("if"))
print(keyword.iskeyword("len"))
#there is not a keyword len in built-ins
```

```python
import builtins
print("print" in dir(builtins))
print("if" in dir(builtins))
# if is a keyword not a built-ins
```
